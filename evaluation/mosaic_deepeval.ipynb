## Step 4 — Define GroqJudge (async-safe)
Uses AsyncGroq for a_generate so DeepEval's async runner works correctly.

```python
from deepeval.models import DeepEvalBaseLLM
from groq import Groq, AsyncGroq
import asyncio, time

class GroqJudge(DeepEvalBaseLLM):
    def __init__(self):
        self.client       = Groq(api_key=os.environ['GROQ_API_KEY'])
        self.async_client = AsyncGroq(api_key=os.environ['GROQ_API_KEY'])
        self.model        = 'llama-3.3-70b-versatile'

    def load_model(self):
        return self.client

    def generate(self, prompt: str) -> str:
        for attempt in range(5):
            try:
                response = self.client.chat.completions.create(
                    model=self.model,
                    messages=[
                        {'role': 'system', 'content': 'You must respond ONLY with valid JSON. No explanation, no markdown fences, no extra text.'},
                        {'role': 'user',   'content': prompt}
                    ],
                    temperature=0
                )
                return response.choices[0].message.content
            except Exception as e:
                if '429' in str(e) or 'rate_limit' in str(e):
                    wait = 10 * (attempt + 1)
                    print(f'Rate limit hit, waiting {wait}s...')
                    time.sleep(wait)
                else:
                    raise
        raise RuntimeError('Max retries exceeded')

    async def a_generate(self, prompt: str) -> str:
        for attempt in range(5):
            try:
                response = await self.async_client.chat.completions.create(
                    model=self.model,
                    messages=[
                        {'role': 'system', 'content': 'You must respond ONLY with valid JSON. No explanation, no markdown fences, no extra text.'},
                        {'role': 'user',   'content': prompt}
                    ],
                    temperature=0
                )
                return response.choices[0].message.content
            except Exception as e:
                if '429' in str(e) or 'rate_limit' in str(e):
                    wait = 10 * (attempt + 1)
                    print(f'Rate limit hit, waiting {wait}s...')
                    await asyncio.sleep(wait)
                else:
                    raise
        raise RuntimeError('Max retries exceeded')

    def get_model_name(self) -> str:
        return self.model

print('GroqJudge defined.')
```

## Step 6 — Fetch retrieval context from Pinecone
Now that Pinecone is populated, this returns real document chunks.

```python
from pinecone import Pinecone
from sentence_transformers import SentenceTransformer

print('Loading BGE embedder...')
embedder = SentenceTransformer('BAAI/bge-small-en-v1.5')

pc    = Pinecone(api_key=os.environ['PINECONE_API_KEY'])
index = pc.Index(PINECONE_INDEX_NAME)

# Check index is populated
stats = index.describe_index_stats()
total_vectors = stats.get('total_vector_count', 0)
print(f'Pinecone index: {PINECONE_INDEX_NAME} — {total_vectors} vectors')
if total_vectors == 0:
    print('WARNING: Pinecone is still empty — re-run the ingestion cell above!')
else:
    print('Pinecone populated — retrieval context will be real.')

def get_retrieval_context(query: str, top_k: int = 3) -> list:
    prefixed  = f'Represent this sentence for searching relevant passages: {query}'
    embedding = embedder.encode(prefixed).tolist()
    results   = index.query(
        vector=embedding,
        top_k=top_k,
        namespace='knowledge_base',
        include_metadata=True
    )
    chunks = [match.get('metadata', {}).get('text', '') for match in results['matches']]
    return [c for c in chunks if c] or ['No context retrieved']

print('Fetching retrieval context for all 10 questions...')
for tc in raw_test_cases:
    tc['retrieval_context'] = get_retrieval_context(tc['input'])
    preview = tc['retrieval_context'][0][:80] if tc['retrieval_context'] else 'empty'
    print(f"  ✓ {tc['input'][:50]:50s} | context: {preview}...")

print('\nAll retrieval contexts filled.')
```

## Step 7 — Define metrics and build test cases

```python
from deepeval.test_case import LLMTestCase
from deepeval.metrics import (
    FaithfulnessMetric,
    AnswerRelevancyMetric,
    ContextualPrecisionMetric,
    ContextualRecallMetric,
    ContextualRelevancyMetric
)

judge = GroqJudge()

# Using 3 test cases only to stay within Groq free tier (100k tokens/day)
# Increase to 10 if you have a paid Groq plan
N_CASES = 3

faithfulness     = FaithfulnessMetric(threshold=0.7,        model=judge, include_reason=False)
answer_relevancy = AnswerRelevancyMetric(threshold=0.7,     model=judge, include_reason=False)
ctx_precision    = ContextualPrecisionMetric(threshold=0.7, model=judge, include_reason=False)
ctx_recall       = ContextualRecallMetric(threshold=0.7,    model=judge, include_reason=False)
ctx_relevancy    = ContextualRelevancyMetric(threshold=0.7, model=judge, include_reason=False)

mosaic_test_cases = [
    LLMTestCase(
        input=tc['input'],
        actual_output=tc['actual_output'],
        retrieval_context=tc['retrieval_context'],
        expected_output=tc['expected_output']
    )
    for tc in raw_test_cases[:N_CASES]  # only first 3 cases
]

print(f'Running with {len(mosaic_test_cases)} test cases (token-saving mode).')
print('Each metric call uses ~2-5k tokens. 5 metrics x 3 cases = ~75k tokens total.')
```

## Step 8 — Run MOSAICurriculum evaluation

```python
from deepeval import evaluate
from deepeval.evaluate import AsyncConfig
import time

# Run one metric at a time with sleep between to avoid rate limits
print('Running MOSAICurriculum evaluation (one metric at a time)...')
metrics_list = [faithfulness, answer_relevancy, ctx_precision, ctx_recall, ctx_relevancy]
metric_names_list = ['Faithfulness', 'Answer Relevancy', 'Contextual Precision', 'Contextual Recall', 'Contextual Relevancy']

mosaic_scores = {}
for metric, name in zip(metrics_list, metric_names_list):
    print(f'  Evaluating {name}...')
    result = evaluate(
        test_cases=mosaic_test_cases,
        metrics=[metric],
        async_config=AsyncConfig(run_async=False)
    )
    scores = []
    for tr in result.test_results:
        for md in tr.metrics_data:
            scores.append(md.score or 0)
    mosaic_scores[name] = round(sum(scores)/len(scores), 3) if scores else 0.0
    print(f'    {name}: {mosaic_scores[name]}')
    print('  Sleeping 30s to avoid rate limit...')
    time.sleep(30)

print('MOSAICurriculum evaluation complete.')
print(mosaic_scores)
```

## Step 9 — Generate baseline and evaluate

```python
from groq import Groq
import time

groq_client = Groq(api_key=os.environ['GROQ_API_KEY'])

print('Generating baseline responses...')
baseline_test_cases = []
for tc in raw_test_cases[:N_CASES]:
    response = groq_client.chat.completions.create(
        model='llama-3.3-70b-versatile',
        messages=[{'role': 'user', 'content': tc['input']}]
    )
    baseline_output = response.choices[0].message.content
    baseline_test_cases.append(
        LLMTestCase(
            input=tc['input'],
            actual_output=baseline_output,
            retrieval_context=tc['retrieval_context'],
            expected_output=tc['expected_output']
        )
    )
    print(f"  ✓ {tc['input'][:55]}...")
    time.sleep(5)  # small sleep between baseline calls

print('\nRunning baseline evaluation (one metric at a time)...')
baseline_scores = {}
for metric, name in zip(metrics_list, metric_names_list):
    print(f'  Evaluating {name}...')
    result = evaluate(
        test_cases=baseline_test_cases,
        metrics=[metric],
        async_config=AsyncConfig(run_async=False)
    )
    scores = []
    for tr in result.test_results:
        for md in tr.metrics_data:
            scores.append(md.score or 0)
    baseline_scores[name] = round(sum(scores)/len(scores), 3) if scores else 0.0
    print(f'    {name}: {baseline_scores[name]}')
    print('  Sleeping 30s to avoid rate limit...')
    time.sleep(30)

print('Baseline evaluation complete.')
print(baseline_scores)
```

## Step 10 — Results table

```python
import pandas as pd

metric_names = ['Faithfulness', 'Answer Relevancy', 'Contextual Precision', 'Contextual Recall', 'Contextual Relevancy']

rows = []
for name in metric_names:
    m = mosaic_scores.get(name, 0.0)
    b = baseline_scores.get(name, 0.0)
    rows.append({
        'Metric':           name,
        'MOSAICurriculum':  m,
        'Baseline LLM':     b,
        'Delta (MOSAIC-B)': round(m - b, 3)
    })

df = pd.DataFrame(rows)
print(df.to_string(index=False))
print(f'\nN = {N_CASES} test cases')
print('\n── Copy these numbers into your Results section ──')
```
